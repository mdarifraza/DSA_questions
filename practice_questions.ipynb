```python
## 9. print all factor of a given numbers:
from math import sqrt
n=25
num=n
output=[]
for i in range(1,int(sqrt(num))+1):
    if num %i==0:
        output.append(i)
        if num//i!=sqrt(num):
            output.append(num//i)
print(output)
    
```

```python
def Factor(n):
    num=n
    output=[]
    for i in range(1,int(sqrt(num))+1):
        if num%i==0:
            output.append(i)
            if num//i!=sqrt(num):
                output.append(num//i)
    output.sort()
    print(output)
Factor(25)
```

```python
# brute force approach
n=[5,3,2,2,1,5,5,7,5,10]
m=[10,111,1,9,5,6,7,2]

for num in m:
    count=0
    for x in n:
        if x==num:
            count+=1
    print(num, ":", count)
```

```python
#hash_list
n=[5,3,2,2,1,5,5,7,5,10]
m=[10,111,1,9,5,67,2]
hash_list=[0]*11
for num in n:
    hash_list[num]+=1
for num in m:
    if num<1 or num>10:
        print(0)
    else:
        print(num,":",hash_list[num])
```

```python
#using dictionary
n=[5,3,2,2,1,5,5,7,5,10]
m=[10,111,1,9,5,67,2]
frequency_map=dict()
for num in n:
    if num in frequency_map:
        frequency_map[num] += 1
    else:
        frequency_map[num] = 1

for num in m:
    if num in frequency_map:
        print(num, "->", frequency_map[num])
    else:
        print(num, "->", 0)
```

```python
# character_hashing
s="azyxyyzaaaa"
q=["d","a","y","x"]
hash_list=[0]*26
for ch in s:
    ascii_val=ord(ch)
    index=ascii_val-97
    hash_list[index]+=1
for ch in q:
    ascii_val=ord(ch)
    index=ascii_val-97
    print(hash_list[index])
```

```python
s = "azyxyyzaaaa"
q = ["d", "a", "y", "x"]

# Create hash array for 26 lowercase letters
hash_list = [0] * 26

# Store frequency of each character
for ch in s:
    index = ord(ch) - ord('a')
    hash_list[index] += 1

# Find frequency of query characters
for ch in q:
    index = ord(ch) - ord('a')
    print(hash_list[index])
```

```python
# first non repating character
class Solution:
    def firstNonRepeatingCharacter(self, s: str) -> int:
        char = {}

        # Count frequency
        for ch in s:
            char[ch] = char.get(ch, 0) + 1

        # Find first unique character
        for i in range(len(s)):
            if char[s[i]] == 1:
                return s[i]

        return -1


s = "swiss"

obj1 = Solution()

print(obj1.firstNonRepeatingCharacter(s))
```
