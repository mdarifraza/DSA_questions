```python
# check if the number is palindrome or not
n=1234
num=n
result=0
while num>n:
    last_digit=num%10
    result=(result*10)+last_digit
    num=num//10
    print(result)
if n==result:
    print("palindrome")
else:
    print("Not palindrome")
```

```python
def check_palindrome(n):
    num=n
    result=0
    while num>0:
        last_digit=num%10
        result=(result*10)+last_digit
        num=num//10
        #print(result)
    if n==result:
        print("palindrome")
    else:
        print("Not palindrome")
n=121
check_palindrome(n)
```

```python
## 8. Check if the Armstrong number
n=153
num=n
total=0
num_of_digits=len(str(num))
while num>0:
    last_digit=num%10
    total=total+(last_digit**num_of_digits)
    num=num//10
    print(total)
if n==total:
    print("its a Armstrong")
else:
    print("not Armstrong")
```
